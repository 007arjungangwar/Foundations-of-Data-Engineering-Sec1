# **Books scraping multiple page** (1000 results - showing 1 to 20.)

 Loop through the first 10 pages


```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

all_books_data = []

# Loop through the first 10 pages
for page_num in range(1, 11):
    url = f"http://books.toscrape.com/catalogue/page-{page_num}.html"
    print(f"Scraping page {page_num}: {url}")

    response = requests.get(url)
    if response.status_code != 200:
        print(f"Page {page_num} not found. Stopping.")
        break

    soup = BeautifulSoup(response.content, "html.parser")
    all_books_on_page = soup.find_all("article", class_="product_pod")

    for book in all_books_on_page:
        title = book.h3.a["title"]
        price = book.find("p", class_="price_color").text.strip("£")
        rating = book.p["class"][1]

        all_books_data.append({
            "title": title,
            "price": price,
            "rating": rating
        })

    time.sleep(1)
```

```python
# Convert list to DataFrame
df = pd.DataFrame(all_books_data)
```

# **Adding the image url**

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

site_url = "http://books.toscrape.com/"
all_books_data = []

for page_num in range(1, 6):
    url = f"http://books.toscrape.com/catalogue/page-{page_num}.html"
    response = requests.get(url)

    if response.status_code != 200:
        break

    soup = BeautifulSoup(response.content, "html.parser")
    all_books_on_page = soup.find_all("article", class_="product_pod")

    for book in all_books_on_page:
        title = book.h3.a["title"]
        price = book.find("p", class_="price_color").text.strip("£")
        rating = book.p["class"][1]

        # Extract relative src attribute (e.g., 'media/cache/...')
        img_src = book.find("div", class_="image_container").find("img")["src"]

        # Convert relative path to absolute URL
        img_url = img_src.replace("../", "")

        all_books_data.append({
            "title": title,
            "price": price,
            "rating": rating,
            "image_url": img_url
        })

    time.sleep(1)
```

# **Add stock availability**

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import pandas as pd
import time

BASE_SITE = "http://books.toscrape.com/"
all_books_data = []

for page_num in range(1, 6):
    url = f"http://books.toscrape.com/catalogue/page-{page_num}.html"
    response = requests.get(url)

    if response.status_code != 200:
        break

    soup = BeautifulSoup(response.content, "html.parser")
    all_books_on_page = soup.find_all("article", class_="product_pod")

    for book in all_books_on_page:
        title = book.h3.a["title"]
        price = book.find("p", class_="price_color").text.strip("£")
        rating = book.p["class"][1]

        # Extract relative src attribute (e.g., 'media/cache/...')
        img_src = book.find("div", class_="image_container").find("img")["src"]

        # Convert relative path to absolute URL
        full_img_url = urljoin(BASE_SITE, img_src.replace("../", ""))

        # Add stock availability
        stock_avail=book.find("p", class_="instock availability").text.strip()
        all_books_data.append({
            "title": title,
            "price": price,
            "rating": rating,
            "image_url": full_img_url,
            "stock availability" : stock_avail
        })

    time.sleep(1)
```

```python
# Convert to DataFrame and export to CSV
df = pd.DataFrame(all_books_data)
df.head()
```

# **Using BeautifulSoup**

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# 1. Define travel dates
checkin_date = '2026-09-01'
checkout_date = '2026-09-02'

page_url = (
    f'https://www.booking.com/searchresults.en-us.html?'
    f'checkin={checkin_date}&checkout={checkout_date}&'
    f'selected_currency=USD&ss=Paris&ssne=Paris&ssne_untouched=Paris&'
    f'lang=en-us&sb=1&src_elem=sb&src=searchresults&dest_type=city&'
    f'group_adults=1&no_rooms=1&group_children=0&sb_travel_purpose=leisure'
)

# 2. Set headers to mimic a real browser request
headers = {
    'User-Agent': (
        'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
        'AppleWebKit/537.36 (KHTML, like Gecko) '
        'Chrome/120.0.0.0 Safari/537.36'
    ),
    'Accept-Language': 'en-US,en;q=0.9',
}
```

```python
# 3. Send HTTP request
response = requests.get(page_url, headers=headers)
print(f"Status Code: {response.status_code}")

soup = BeautifulSoup(response.content, 'html.parser')

# 4. Find property cards
hotels = soup.find_all('div', {'data-testid': 'property-card'})
print(f'There are: {len(hotels)} hotels found.')
```

```python
hotels_list = []

# 5. Extract hotel details
for hotel in hotels:
    hotel_dict = {}

    # Title
    title_tag = hotel.find('div', {'data-testid': 'title'})
    hotel_dict['hotel'] = title_tag.get_text(strip=True) if title_tag else None

    # Price
    price_tag = hotel.find('span', {'data-testid': 'price-and-discounted-price'})
    hotel_dict['price'] = price_tag.get_text(strip=True) if price_tag else None

    # Rating and Reviews
    review_container = hotel.find('div', {'data-testid': 'review-score'})
    if review_container:
        score_tag = review_container.find('div')
        hotel_dict['score'] = score_tag.get_text(strip=True) if score_tag else None

        sub_divs = review_container.find_all('div', recursive=False)
        if len(sub_divs) > 1:
            labels = sub_divs[1].find_all('div')
            hotel_dict['avg review'] = labels[0].get_text(strip=True) if len(labels) > 0 else None

            if len(labels) > 1:
                raw_reviews = labels[1].get_text(strip=True)
                hotel_dict['reviews count'] = raw_reviews.split()[0] if raw_reviews else None
            else:
                hotel_dict['reviews count'] = None
        else:
            hotel_dict['avg review'] = None
            hotel_dict['reviews count'] = None
    else:
        hotel_dict['score'] = None
        hotel_dict['avg review'] = None
        hotel_dict['reviews count'] = None

    hotels_list.append(hotel_dict)
```

```python
# 6. Create DataFrame & Save
df = pd.DataFrame(hotels_list)
```

```python
df.to_excel('hotels_list.xlsx', index=False)
df.to_csv('hotels_list.csv', index=False)
```
