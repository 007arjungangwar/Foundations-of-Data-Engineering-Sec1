````python
# Imports & Base Text
import nlpaug.augmenter.word as naw
````

````python
text = 'I have a tree, a green, green tree,To shade me from the sun;And under it I often sit,When all my play is done.'
````

````python
import nltk
nltk.download('averaged_perceptron_tagger_eng')
````

# Synonym Replacement

````python
# Synonym Replacement
syn_aug = naw.SynonymAug(aug_src='wordnet')
synonym_text = syn_aug.augment(text)
print('Original Text:', text)
print('Synonym Text:', synonym_text)
````

# Antonym Replacement

````python
# antonym Replacement
aug = naw.AntonymAug()
antonym_text = aug.augment(text)
print('Original Text:', text)
print('antonym Text:', antonym_text)
````

# Random Substitution

````python
# Random Substitution
aug = naw.RandomWordAug(action='substitute',target_words=["Arjun","Gangwar","Rohan"])
substituted_text = aug.augment(text)
print('Original Text:', text)
print('Substituted Text:', substituted_text)
````

````python
# Random Substitution
aug = naw.RandomWordAug(action='substitute',target_words=["###","@@@","***"])
substituted_text = aug.augment(text)
print('Original Text:', text)
print('Substituted Text:', substituted_text)
````

# Random Deletion

````python
# Random Deletion
del_aug = naw.RandomWordAug(action='delete')
deletion_text = del_aug.augment(text)
print('Original Text:', text)
print('Deletion Text:', deletion_text)
````

# Random Swap

````python
# Random Swap
swap_aug = naw.RandomWordAug(action='swap',target_words=["Arjun","Gangwar","Rohan"])
swap_text = swap_aug.augment(text)
print('Original Text:', text)
print('Swap Text:', swap_text)
````

# [**Reserved words with key and value**]

````python
aug = naw.ReservedAug(reserved_tokens=[['tree','Rohan'], ['green','red'],['sun','Arjun']])
augmented_data = aug.augment(text)
print('Original Text:', text)
print('Reserved Text:', augmented_data)
````

# translate original text to other language (german) and convert back to english language

````python
# translate original text to other language (german) and convert back to english language
back_trans_aug = naw.back_translation.BackTranslationAug()
back_trans_text = back_trans_aug.augment(text)
print('Original Text:', text)
print('Back Translated Text:', back_trans_text)
````

# **word_embs**

# **One sentence and showing all techniques side-by-side:**



```
Original text → Synonym → Substitution → Deletion → Swap → Insertion
→ Word Embedding → Character-level changes → Back Translation
```



[nlpgaug library for textual augmentation in machine learning experiments.](https://nlpaug.readthedocs.io/en/latest/)


[Word Augmenter](https://nlpaug.readthedocs.io/en/latest/augmenter/word/word.html)

````python
import nlpaug.augmenter.word as naw
import nlpaug.augmenter.char as nac
````

````python
text = "The quick brown fox jumps over a lazy dog"
````

````python
# 1. Synonym Replacement
syn_aug = naw.SynonymAug(aug_src='wordnet')
print("Original   :", text)
print("Synonym    :", syn_aug.augment(text))
````

````python
# 2. Random Substitution
sub_aug = naw.RandomWordAug(action='substitute')
print("Substitution:", sub_aug.augment(text))
````

````python
# 3. Random Deletion
del_aug = naw.RandomWordAug(action='delete')
print("Deletion   :", del_aug.augment(text))
````

````python
# 4. Random Swap
swap_aug = naw.RandomWordAug(action='swap')
print("Swap       :", swap_aug.augment(text))
````

````python
# 6. Character Insertion
char_ins = nac.RandomCharAug(action='insert')
print("Char Insert:", char_ins.augment(text))
````

````python
# 7. Character Deletion
char_del = nac.RandomCharAug(action='delete')
print("Char Delete:", char_del.augment(text))
````

````python
# 8. Character Swap
char_swap = nac.RandomCharAug(action='swap')
print("Char Swap  :", char_swap.augment(text))
````
