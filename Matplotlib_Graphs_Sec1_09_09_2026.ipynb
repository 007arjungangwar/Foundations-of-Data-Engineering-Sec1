```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv(r"Titanic-Dataset.csv")
df.head(2)
```

```python
gender=df["Sex"].value_counts()
gender
```

```python
type(gender)
```

```python
gender.index
```

```python
gender.values
```

## **Bar**

```python
plt.figure(figsize=(5,2))
plt.bar(x=gender.index,height=gender.values,color=('Red','green'))
plt.title("Total male and female in the journey")
plt.xlabel("Gender")
plt.ylabel("Count of persons")
plt.show()
```

## **Barh**

```python
plt.figure(figsize=(5,2))
plt.barh(y=gender.index,width=gender.values,color=('Red','green'))
plt.title("Total male and female in the journey")
plt.xlabel("Gender")
plt.ylabel("Count of persons")
plt.show()
```

## **scatter**

```python
df["Age"]=df["Age"].fillna(0)
```

```python
df["Age"]
```

```python
df["Age"].index
```

```python
df["Age"].values
```

```python
plt.figure(figsize=(4,4))
plt.scatter(df["Age"].index,df["Age"].values,s=12,c="red",edgecolors="black")
plt.title("Scatter graph of Age")
plt.xlabel("Index")
plt.ylabel("Age")
plt.show()
```

## **Hist**

```python
plt.figure(figsize=(4,3))
plt.hist(df["Age"].values,bins=20,color="red",alpha=0.5)  #,density=True
plt.title("Histogram of Age")
plt.xlabel("Age")
plt.ylabel("Count of persons")
plt.legend(["Age"])   # plt.axis([0, 80, 0, 250])
plt.show()
```

## **Pie**

```python
df["Embarked"].value_counts()
```

```python
df["Embarked"].value_counts().index
```

```python
df["Embarked"].value_counts().values
```

```python
plt.figure(figsize=(4,3))
plt.pie(df["Embarked"].value_counts().values,explode=(0.1,0.1,0.1),autopct="%.2f%%",radius=0.9,center=(0,0))
plt.legend(labels=df["Embarked"].value_counts().index,loc="best")
plt.title("Percentage Of Passenger Count where they took the ship From the place(Embarked)")
plt.show()
```

```python
df['Pclass'].value_counts()
```

```python
plt.figure(figsize=(4,3))
plt.plot(df['Pclass'].value_counts().index,df['Pclass'].value_counts().values,'-',linewidth=2.0,color='r',marker="o",linestyle="--")
plt.title("Total Number of passenger based on their passenger class")
plt.xlabel("Passenger Class")
plt.ylabel("Count of persons")
plt.show()
```

```python
df["total_relative"]=df["SibSp"]+df["Parch"]
df["total_relative"]
```

```python
df["total_relative"].value_counts()
```

```python
plt.figure(figsize=(4,3))
plt.plot(df["total_relative"].value_counts().index,df["total_relative"].value_counts().values,'-',linewidth=2.0)
plt.title("Total Number of passenger based on their total relative")
plt.xlabel("Total relative")
plt.ylabel("Count of persons")
plt.show()
```
