## **Data visualization Through the matplotlib**

```python
import matplotlib.pyplot as plt
```

```python
df.dtypes
```

```python
plt.figure(figsize=(8,4))

plt.subplot(1,2,1)
plt.hist(df["Age"].dropna(), bins=20)
plt.title("AgeDistribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.subplot(1,2,2)
plt.hist(df["Fare"], bins=30)
plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")

plt.show()
```

```python
df["Age"].value_counts().index
```

```python
df["Age"].value_counts().values
```

```python
df["Fare"].values
```

```python
df["Sex"]
```

```python
plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.hist(df["Age"].value_counts().values,bins=20)

plt.subplot(2,2,2)
plt.hist(df["Fare"].values,bins=15)

plt.subplot(2,2,3)
plt.scatter(df["Fare"].index,df["Fare"].values,c="Red")

plt.subplot(2,2,4)
plt.barh(df["Sex"].value_counts().index,df["Sex"].value_counts().values,color=["Red","Orange"])
plt.show()
```

```python
fig,axes=plt.subplots(2,2)

axes[0,0].hist(df["Age"].value_counts().values,bins=20)
axes[0,1].hist(df["Fare"].values,bins=15)
axes[1,0].scatter(df["Fare"].index,df["Fare"].values,c="Red")
axes[1,1].barh(df["Sex"].value_counts().index,df["Sex"].value_counts().values,color=["Red","Orange"])
plt.tight_layout()
plt.show()
```

## **Data visualization Through the Seaborn**

```python
import seaborn as sns
import matplotlib.pyplot as plt

fig, axes =plt.subplots(2,2,figsize=(12,8))

sns.histplot(data=df,x="Age",bins=20,ax=axes[0, 0],kde=True)
axes[0, 0].set_title("Age Distribution")

sns.histplot(data=df,x="Fare",bins=30,ax=axes[0, 1],kde=True)
axes[0, 1].set_title("Fare Distribution")

sns.countplot(data=df,x="Sex",ax=axes[1, 0],hue="Sex",palette=["Red","Orange"],legend=True)
axes[1, 0].set_title("Passenger Count byGender")

sns.countplot(data=df,x="Pclass",ax=axes[1, 1],hue="Pclass",palette=["Red","Orange","Green"],legend=True)
axes[1, 1].set_title("Passenger Count byClass")

plt.tight_layout()
plt.show()
```
